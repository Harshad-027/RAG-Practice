```python
sample_texts = {
    "../data/text_files/python_intro.txt": """
Python is a high-level, general-purpose programming language known for its
clear syntax and readability. It is widely used in web development,
automation, data analysis, artificial intelligence, scientific computing,
and software testing. Python programs are usually shorter and easier to
understand than equivalent programs written in many lower-level languages.
Its large standard library provides useful modules for tasks such as working
with files, dates, mathematics, regular expressions, and networking.
Python supports multiple programming styles, including procedural,
object-oriented, and functional programming. Variables do not require
explicit type declarations, which makes rapid experimentation convenient.
Popular libraries such as NumPy, pandas, Flask, Django, and Matplotlib
extend Python’s capabilities. Beginners often choose Python because its
simple structure helps them focus on logic, problem-solving, and building
practical projects rather than complex language syntax.
""".strip(),

    "../data/text_files/machine_learning.txt": """
Machine learning is a branch of artificial intelligence in which computer
systems learn patterns from data instead of relying only on manually written
rules. A machine-learning model is trained using examples, then used to make
predictions or decisions on new, unseen data. Common applications include
spam-email detection, product recommendations, image recognition, fraud
detection, language translation, and medical-data analysis. Supervised
learning uses labeled data, such as emails marked spam or not spam.
Unsupervised learning finds hidden groups or patterns in unlabeled data.
Reinforcement learning improves decisions through rewards and penalties.
The quality of data, feature selection, model choice, and evaluation methods
strongly affect results. Python is commonly used for machine learning because
libraries such as scikit-learn, TensorFlow, PyTorch, NumPy, and pandas make
data preparation, training, testing, and deployment more practical.
""".strip()
}





for filepath, content in sample_texts.items():
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(content)

print('Sample text files created !! ')
```

```python
### TextLoader 

from langchain_community.document_loaders import TextLoader
```

```python
loader = TextLoader('../data/text_files/python_intro.txt', encoding='utf-8')
document=loader.load()
```

```python
print(document)
```

```python
### Directory Loader 

from langchain_community.document_loaders import DirectoryLoader

### used to load all files from the directory

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", ##Pattern to match files
    loader_cls=TextLoader,
    loader_kwargs={'encoding' : 'utf-8'},
    show_progress=False
)
```

```python
documents=dir_loader.load()
documents
```

```python
from langchain_community.document_loaders import PyMuPDFLoader

dir_loader = DirectoryLoader(
    "../data/pdfs/",
    glob="**/*.pdf", ##Pattern to match files
    loader_cls=PyMuPDFLoader,
    show_progress=False
)
```

```python
pdfs = dir_loader.load()
pdfs
```
